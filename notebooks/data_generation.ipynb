```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta
```

```python
np.random.seed(42)
```

```python
# -----------------------
# CONFIG
# -----------------------
NUM_USERS = 10000
NUM_TRANSACTIONS = 200000

countries = ["India"]
cities = ["Delhi", "Mumbai", "Indore", "Pune", "Jammu"]
channels = ["ads", "organic", "referral"]
payment_methods = ["upi", "card", "wallet"]
failure_reasons = ["insufficient_balance", "network_error", "bank_declined", "timeout"]
```

```python
# -----------------------
# USERS TABLE
# -----------------------
users = pd.DataFrame({
    "user_id": range(1, NUM_USERS + 1),
    "signup_date": pd.to_datetime("2022-01-01") + pd.to_timedelta(np.random.randint(0, 900, NUM_USERS), unit="D"),
    "country": np.random.choice(countries, NUM_USERS),
    "city": np.random.choice(cities, NUM_USERS),
    "acquisition_channel": np.random.choice(channels, NUM_USERS, p=[0.5, 0.3, 0.2])
})

users.head()
```

```python
# -----------------------
# TRANSACTIONS TABLE
# -----------------------
transactions = pd.DataFrame({
    "transaction_id": range(1, NUM_TRANSACTIONS + 1),
    "user_id": np.random.randint(1, NUM_USERS + 1, NUM_TRANSACTIONS),
    "transaction_date": pd.to_datetime("2023-01-01") + pd.to_timedelta(np.random.randint(0, 365, NUM_TRANSACTIONS), unit="D"),
    "amount": np.round(np.random.exponential(scale=2000, size=NUM_TRANSACTIONS), 2),
    "status": np.random.choice(["success", "failed"], NUM_TRANSACTIONS, p=[0.88, 0.12]),
    "payment_method": np.random.choice(payment_methods, NUM_TRANSACTIONS)
})

transactions
```

```python
# -----------------------
# PAYOUTS TABLE (ONLY SUCCESS TRANSACTIONS)
# -----------------------
success_tx = transactions[transactions["status"] == "success"].copy()

payouts = pd.DataFrame({
    "payout_id": range(1, len(success_tx) + 1),
    "transaction_id": success_tx["transaction_id"],
    "payout_amount": success_tx["amount"] * np.random.uniform(0.90, 0.98, len(success_tx)),
    "commission_fee": success_tx["amount"] * np.random.uniform(0.02, 0.10, len(success_tx)),
    "payout_status": np.random.choice(["completed", "pending", "failed"], len(success_tx), p=[0.85, 0.10, 0.05])
})

payouts
```

```python
# -----------------------
# FAILURE LOGS TABLE (ONLY FAILED TRANSACTIONS)
# -----------------------
failed_tx = transactions[transactions["status"] == "failed"].copy()

failure_logs = pd.DataFrame({
    "failure_id": range(1, len(failed_tx) + 1),
    "transaction_id": failed_tx["transaction_id"],
    "failure_reason": np.random.choice(failure_reasons, len(failed_tx)),
    "failure_time": failed_tx["transaction_date"] + pd.to_timedelta(np.random.randint(0, 60, len(failed_tx)), unit="m")
})

failure_logs
```

```python
# -----------------------
# SAVE FILES
# -----------------------
users.to_csv("users.csv", index=False)
transactions.to_csv("transactions.csv", index=False)
payouts.to_csv("payouts.csv", index=False)
failure_logs.to_csv("failure_logs.csv", index=False)

print("Data generated successfully!")
```
